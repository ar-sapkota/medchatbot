```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
def chunk_data():
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size = 100,
        chunk_overlap = 20,
        length_function = len
    )
    texts = text_splitter.create_documents([pages])
    return texts
```

```python

chunks = chunk_data()
```

```python
def embedding():
    """download and return huggingface embedding"""
    model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
    texts = [chunk.page_content for chunk in chunks]
    embeddings = model.encode(texts)

    return embeddings
```

#### Creating langchain compatible retriever

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from sentence_transformers import SentenceTransformer
```

```python
embed_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
def create_embeddings():
    '''creates and return embeddings for using in query time'''
    return HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
embeddings = create_embeddings()


def create_vectorStore(embeddings):
    '''create and return pinecone vector store related to the existing index'''
# connect to existing vector store
    vector_store = PineconeVectorStore(
    index_name='langchainvector',
    embedding=embeddings,
    namespace='default' )
    return vector_store
vector_store = create_vectorStore(embeddings) #connect to existing index

def create_retriever(vector_store,top_k=10):
    '''create and return retriever for searching the index'''
    # this retriever searches top 4 matches from indexed chunks
    retriever = vector_store.as_retriever(search_kwargs={'k':top_k})
    return retriever
retriever = create_retriever(vector_store) #for searching
```

#### Chain Creation

```python
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
```

```python
# creation of document chain
def create_doc_chain(llm,prompt):
    '''create and return document chain for combining retrieved docs.'''
    doc_chain = create_stuff_documents_chain(llm,prompt)

    return doc_chain
document_chain = create_doc_chain(llm,prompt)

# create full retrieval chain.
from langchain_classic.chains import create_retrieval_chain
def create_full_retrieval_chain(retriever, document_chain):
    '''create and return full retrieval chain.'''
    ret_chain = create_retrieval_chain(retriever, document_chain)
    return ret_chain
retrieval_chain = create_retrieval_chain(retriever, document_chain)
    
```
